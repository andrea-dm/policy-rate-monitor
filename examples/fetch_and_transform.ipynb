# Fetch and transform

This notebook prepares the data that [policy_rates.ipynb](policy_rates.ipynb) reads: it downloads the BIS file and stages it, as `bis-prates fetch` and `bis-prates transform` do.

## Setup

```python
from pathlib import Path
```

```python
from policy_rate_monitor.bulk import BulkFileProvider
from policy_rate_monitor.config import FileConfigSource
```

The notebook is meant to run from `examples/`, so the repository root, `ROOT`, is its parent folder.  

> 👉 Change `ROOT` if you run the notebook from elsewhere.

```python
ROOT = Path.cwd().parent
```

```python
config = FileConfigSource(ROOT / "configs" / "bis_cbpol.json").load()
provider = BulkFileProvider(config, ROOT / config.dataset.data_dir)
```

## Fetch

Downloads the ZIP from the BIS, unless the cached copy is less than 24 hours old (`cache_ttl_hours` in the configuration).  
The release records where the file came from, when it was retrieved and its fingerprint.

```python
provider.fetch()
```

## Transform

Loads the CSV as it is into `data/raw/`, then stages the daily series into `data/staged/`, with the data-quality findings.

```python
provider.transform()
observations = provider.observations().num_rows
series = len(provider.list_series())
findings = len(provider.data_quality_issues())

print(
	f"{observations} observations of {series} series kept,",
	f"{findings} data-quality findings",
)
```

The staged data is ready: continue with [policy_rates.ipynb](policy_rates.ipynb).
